## 1. Problem Statement

***Estimated number of border crossings at specific U.S. border ports based on features such as port location, month/year, border type, and crossing category.***

Goal: To build a predictive model that accurately estimates the number of border crossings at various U.S. border ports using historical data and relevant features. 

Why: Predicting border crossings in specific locations can help in resource allocation, security planning, and infrastructure development at border ports. Border crossings are a critical aspect of national security and economic activity, and accurate predictions can aid in better management and planning.

Where I got the data: [Border Crossing Entry Data](https://catalog.data.gov/dataset/border-crossing-entry-data-683ae)


## 2. Data Dictionary
| Variable      | Type                | Description                                                                                           |
| ------------- | ------------------- | ----------------------------------------------------------------------------------------------------- |
| **Port Name** | Categorical         | Name of the U.S. land port where the crossing occurred.                                               |
| **State**     | Categorical         | U.S. state where the port is located.                                                                 |
| **Port Code** | Numeric             | Unique numeric identifier assigned to each border port.                                               |
| **Border**    | Categorical         | Specifies US–Canada or US–Mexico border.                                                              |
| **Date**      | Date                | Month and year of the recorded crossing activity.                                                     |
| **Measure**   | Categorical         | Mode or category of crossing (e.g., Trucks, Personal Vehicles, Train Containers, Bus Passengers).     |
| **Value**     | Numeric             | Count of recorded entries for the given measure at the given port.                                    |
| **Latitude**  | Numeric             | Port latitude based on GIS data.                                                                      |
| **Longitude** | Numeric             | Port longitude based on GIS data.                                                                     |
| **Point**     | Text (WKT Geometry) | Combined latitude/longitude coordinates in POINT format. |


## 3. Data Imports

```python
import pandas as pd
df = pd.read_csv("Border_Crossing_Entry_Data.csv")
df.head()
```

## 4. Data Cleaning

```python
df.isnull().sum()
```

```python
df = df.drop(columns=['Point']) #dropping 'Point' column as theres no use of it in prediction
df['State'] = df['State'].fillna('Unknown') #filling missing values in 'State' with 'Unknown'
df['Latitude'] = df['Latitude'].fillna(df['Latitude'].median()) #filling missing values in 'Latitude' with median
df['Longitude'] = df['Longitude'].fillna(df['Longitude'].median()) #filling missing values in 'Longitude' with median
df.isnull().sum()
```

```python
#date formatting
df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
```

```python
df.drop_duplicates(inplace=True)#removing duplicates
```

```python
df['Measure'] = df['Measure'].str.strip()#removing leading and trailing spaces
df['Port Name'] = df['Port Name'].str.strip()
```
